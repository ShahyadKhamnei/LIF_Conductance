```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
E_s = np.linspace(1.5,8,101,True)
E_l = (-16*E_s**2+48*E_s-16)/20
plt.plot(E_s, E_l, 'b', label='Esyn-Eleak bifurcation w/ J=4')
E_s = np.linspace(-2,1.5,101,True)
E_l = (-16*E_s**2+48*E_s-16)/20
plt.plot(E_s, E_l, 'b--')
E_s = np.linspace(-2,8,101,True)
plt.plot(E_s, np.ones(len(E_s)), 'g', label='Eleak=1 bifurcation')
plt.annotate('Bistable', (5,0))
plt.annotate('High', (2,1.5))
plt.annotate('Low', (-1.5,0))
plt.scatter([3,3,1], [0.5,1.5,0.5])
plt.vlines(1.5, -4, 1, colors='r', linestyles='dashed', label='bifurcation to ensure v>1 w/ J=4')
plt.xlabel('E_s')
plt.ylabel('E_l')
plt.ylim(-4,4)
plt.legend()
```
